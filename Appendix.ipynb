# SDE 的正向过程与反向过程

我们证明 $$\mathrm{d}\mathbf{x} = [\mathbf{f}(\mathbf{x}, t) - g(t)^2 \nabla_{\mathbf{x}} \log p_t(\mathbf{x})]\mathrm{d}t + g(t)\mathrm{d}\bar{\mathbf{w}}$$
上述 SDE 对应了逆向过程。 $$\mathrm{d}\mathbf{x} = \mathbf{f}(\mathbf{x}, t)\mathrm{d}t + g(t)\mathrm{d}\mathbf{w}$$
上述 SDE 对应了正向过程。

对于任何一个标准 SDE $d\mathbf{x} = \mathbf{f}(\mathbf{x}, t)dt + g(t)d\mathbf{w}$，它对应的概率密度 $p_t(\mathbf{x})$ 的演化由 Fokker-Planck 方程描述$$\frac{\partial p_t(\mathbf{x})}{\partial t} = -\nabla \cdot (\mathbf{f}(\mathbf{x}, t)p_t(\mathbf{x})) + \frac{1}{2}g(t)^2 \nabla^2 p_t(\mathbf{x})$$

待证。

# 分数函数工程估计

我们证明一件事，最小化对已知原图条件得分的估计误差，在数学上等价于最小化对全分布边缘得分的估计误差。

我们想要的边缘分布$$L_{explicit} = \mathbb{E}_{p_t(\mathbf{x})} [ \| \mathbf{s}_\theta(\mathbf{x}, t) - \nabla_{\mathbf{x}} \log p_t(\mathbf{x}) \|^2 ]$$
实际上的条件得分$$L_{denoising} = \mathbb{E}_{p(\mathbf{x}_0) p_t(\mathbf{x}|\mathbf{x}_0)} [ \| \mathbf{s}_\theta(\mathbf{x}, t) - \nabla_{\mathbf{x}} \log p_t(\mathbf{x}|\mathbf{x}_0) \|^2 ]$$
我们要证明最小化 $L_{denoising}$ 等价于最小化 $L_{explicit}$。

我们将 $L_{denoising}$ 展开：$$L_{denoising} = \mathbb{E}_{p(\mathbf{x}_0, \mathbf{x})} [ \| \mathbf{s}_\theta \|^2 - 2 \mathbf{s}_\theta \cdot \nabla_{\mathbf{x}} \log p_t(\mathbf{x}|\mathbf{x}_0) + \| \nabla_{\mathbf{x}} \log p_t(\mathbf{x}|\mathbf{x}_0) \|^2 ]$$其中最后一项与 $\theta$ 无关。我们重点关注中间的交叉项$$I = \mathbb{E}_{p(\mathbf{x}_0, \mathbf{x})} [ \mathbf{s}_\theta(\mathbf{x}, t) \cdot \nabla_{\mathbf{x}} \log p_t(\mathbf{x}|\mathbf{x}_0) ]$$

利用期望的性质 $\mathbb{E}_{p(\mathbf{x}_0, \mathbf{x})} [\cdot] = \mathbb{E}_{p_t(\mathbf{x})} [ \mathbb{E}_{p(\mathbf{x}_0|\mathbf{x})} [\cdot] ]$ $$I = \int p_t(\mathbf{x}) \mathbf{s}_\theta(\mathbf{x}, t) \cdot \left( \int p(\mathbf{x}_0|\mathbf{x}) \nabla_{\mathbf{x}} \log p_t(\mathbf{x}|\mathbf{x}_0) d\mathbf{x}_0 \right) d\mathbf{x}$$

根据贝叶斯定理 $p(\mathbf{x}_0|\mathbf{x}) = \frac{p_t(\mathbf{x}|\mathbf{x}_0) p(\mathbf{x}_0)}{p_t(\mathbf{x})}$ $$\int p(\mathbf{x}_0|\mathbf{x}) \nabla_{\mathbf{x}} \log p_t(\mathbf{x}|\mathbf{x}_0) d\mathbf{x}_0 = \int \frac{p_t(\mathbf{x}|\mathbf{x}_0) p(\mathbf{x}_0)}{p_t(\mathbf{x})} \cdot \frac{\nabla_{\mathbf{x}} p_t(\mathbf{x}|\mathbf{x}_0)}{p_t(\mathbf{x}|\mathbf{x}_0)} d\mathbf{x}_0$$
$$= \frac{1}{p_t(\mathbf{x})} \int p(\mathbf{x}_0) \nabla_{\mathbf{x}} p_t(\mathbf{x}|\mathbf{x}_0) d\mathbf{x}_0$$
$$= \frac{1}{p_t(\mathbf{x})} \nabla_{\mathbf{x}} \left( \int p(\mathbf{x}_0) p_t(\mathbf{x}|\mathbf{x}_0) d\mathbf{x}_0 \right)$$
括号内正是全概率公式定义的 $p_t(\mathbf{x})$ $$= \frac{\nabla_{\mathbf{x}} p_t(\mathbf{x})}{p_t(\mathbf{x})} = \nabla_{\mathbf{x}} \log p_t(\mathbf{x})$$

代回交叉项 $I$ $$I = \int_{\mathbf{x}} p_t(\mathbf{x}) \mathbf{s}_{\theta}(\mathbf{x}, t) \cdot \nabla_{\mathbf{x}} \log p_t(\mathbf{x}) \mathrm{d}\mathbf{x}$$ 这就是 $$I = \mathbb{E}_{p_t(\mathbf{x})} [ \mathbf{s}_{\theta}(\mathbf{x}, t) \cdot \nabla_{\mathbf{x}} \log p_t(\mathbf{x}) ]$$
证明结束了。这与我们需要的一致，实际上上式就是 $L_{explicit}$ 展开后的交叉项。

# 关于转移核与 SDE

我们想要证明，在 DDIM 或 SMLD 语境下$$\frac{\mathrm{d}\boldsymbol{\mu}_t}{\mathrm{d}t} = \mathbf{f}(\boldsymbol{\mu}_t, t)$$
其中 $\boldsymbol{\mu}_t$ 是转移核的高斯分布均值，$f$ 是 SDE 的漂移项系数。

考虑定义扩散过程的正向 Itô SDE $$\mathrm{d}\mathbf{x} = \mathbf{f}(\mathbf{x}, t)\mathrm{d}t + g(t)\mathrm{d}\mathbf{w}$$将其写成积分形式 $$\mathbf{x}(t) = \mathbf{x}(0) + \int_{0}^{t} \mathbf{f}(\mathbf{x}(s), s) \mathrm{d}s + \int_{0}^{t} g(s) \mathrm{d}\mathbf{w}(s)$$我们对等式两边取关于初始条件 $\mathbf{x}(0) = \mathbf{x}_0$ 的条件期望 $E[\cdot | \mathbf{x}_0]$。由于高斯噪声的性质，有 $E[\mathbf{x}(t) | \mathbf{x}_0] = \boldsymbol{\mu}_t$。

利用期望的线性性质，我们将期望算子作用于每一项 $$E[\mathbf{x}(t) | \mathbf{x}_0] = E[\mathbf{x}(0) | \mathbf{x}_0] + E\left[ \int_{0}^{t} \mathbf{f}(\mathbf{x}(s), s) \mathrm{d}s \middle| \mathbf{x}_0 \right] + E\left[ \int_{0}^{t} g(s) \mathrm{d}\mathbf{w}(s) \middle| \mathbf{x}_0 \right]$$第一项 $E[\mathbf{x}(0) | \mathbf{x}_0] = \mathbf{x}_0$。第三项根据 Itô 积分的性质，由于 $\mathrm{d}\mathbf{w}(s)$ 是独立于过去状态的增量，且 $E[\mathrm{d}\mathbf{w}(s)] = 0$，因此随机积分的期望恒为 0 $$E\left[ \int_{0}^{t} g(s) \mathrm{d}\mathbf{w}(s) \middle| \mathbf{x}_0 \right] = 0$$现在等式简化为 $$\boldsymbol{\mu}_t = \mathbf{x}_0 + \int_{0}^{t} E[\mathbf{f}(\mathbf{x}(s), s) | \mathbf{x}_0] \mathrm{d}s$$
对上式两边关于时间 $t$ 求导 $$\frac{\mathrm{d}\boldsymbol{\mu}_t}{\mathrm{d}t} = E[\mathbf{f}(\mathbf{x}(t), t) | \mathbf{x}_0]$$

在 Score-based 框架（如 VP/VE-SDE）中，我们定义的漂移项 $\mathbf{f}(\mathbf{x}, t)$ 都是关于 $\mathbf{x}$ 的仿射函数（Affine Function） $$\mathbf{f}(\mathbf{x}, t) = \mathbf{A}(t)\mathbf{x} + \mathbf{b}(t)$$对于仿射变换，期望号可以穿透 $$E[\mathbf{f}(\mathbf{x}, t)] = E[\mathbf{A}(t)\mathbf{x} + \mathbf{b}(t)] = \mathbf{A}(t)E[\mathbf{x}] + \mathbf{b}(t) = \mathbf{f}(E[\mathbf{x}], t)$$代入 $\boldsymbol{\mu}_t = E[\mathbf{x}(t)]$，即得 $$\frac{\mathrm{d}\boldsymbol{\mu}_t}{\mathrm{d}t} = \mathbf{f}(\boldsymbol{\mu}_t, t)$$

直觉是，虽然单个粒子的运动由于受到 $g(t)\mathrm{d}\mathbf{w}$ 的扰动而充满了不确定性，但整个粒子群的几何中心均值只受到漂移项 $\mathbf{f}$ 的牵引。噪声项 $g(t)\mathrm{d}\mathbf{w}$ 只负责让分布变宽（增加方差），而不负责移动分布的中心。

# 连续性方程

我们现在知道几件事。

矢量场 $v_t$ 与流 $\phi_t$ 的关系为 $\frac{\mathrm{d}}{\mathrm{d}t} \phi_t(x) = v_t(\phi_t(x))$，初始条件 $\phi_0(x) = x$。

概率密度路径 $p_t$ 由初始分布 $p_0$ 经过流映射 $\phi_t$ 得到，即 $p_t = [\phi_t]_* p_0$。

或者我们写开 $$\int_A p_t(x) \mathrm{d}x = \int_{\phi_t^{-1}(A)} p_0(y) \mathrm{d}y$$

我们想要直接推导概率密度路径 $p$ 与矢量场 $v_t$ 的关系。

我们引入测试函数。考虑一个任意的平滑且具有紧支集的测试函数 $h \in C_c^\infty(\mathbb{R}^d)$。我们可以考察分布 $p_t$ 在该函数下的期望值随时间的变化率 $$I(t) = \mathbb{E}_{x \sim p_t}[h(x)] = \int_{\mathbb{R}^d} h(x) p_t(x) \mathrm{d}x$$利用 Push-forward 的定义，我们可以将变量 $x$ 映射回初始时刻的变量 $y$ $$I(t) = \int_{\mathbb{R}^d} h(\phi_t(y)) p_0(y) \mathrm{d}y$$

对 $I(t)$ 关于时间 $t$ 求导。由于积分区域与 $t$ 无关，且被积函数关于 $t$ 平滑，导数算子可以进入积分号内部 $$\frac{\mathrm{d}I(t)}{\mathrm{d}t} = \int_{\mathbb{R}^d} \frac{\partial}{\partial t} [h(\phi_t(y))] p_0(y) \mathrm{d}y$$根据全导数链式法则 $$\frac{\partial}{\partial t} h(\phi_t(y)) = \nabla h(\phi_t(y)) \cdot \frac{\mathrm{d}\phi_t(y)}{\mathrm{d}t}$$代入已知的 ODE 关系 $\frac{\mathrm{d}}{\mathrm{d}t} \phi_t(y) = v_t(\phi_t(y))$ $$\frac{\mathrm{d}I(t)}{\mathrm{d}t} = \int_{\mathbb{R}^d} [\nabla h(\phi_t(y)) \cdot v_t(\phi_t(y))] p_0(y) \mathrm{d}y$$

利用 Push-forward 关系将积分变量从初始空间的 $y$ 重新变换回时刻 $t$ 空间的 $x$ $$\frac{\mathrm{d}I(t)}{\mathrm{d}t} = \int_{\mathbb{R}^d} [\nabla h(x) \cdot v_t(x)] p_t(x) \mathrm{d}x$$同时，从 $I(t)$ 的原始积分定义出发，其时间导数也可以表示为 $$\frac{\mathrm{d}I(t)}{\mathrm{d}t} = \int_{\mathbb{R}^d} h(x) \frac{\partial p_t(x)}{\partial t} \mathrm{d}x$$因此，我们得到以下等式 $$\int_{\mathbb{R}^d} h(x) \frac{\partial p_t(x)}{\partial t} \mathrm{d}x = \int_{\mathbb{R}^d} p_t(x) v_t(x) \cdot \nabla h(x) \mathrm{d}x$$

对等式右侧应用分部积分。利用恒等式 $\nabla \cdot (h p_t v_t) = h \nabla \cdot (p_t v_t) + p_t v_t \cdot \nabla h$，可得 $$\int_{\mathbb{R}^d} p_t v_t \cdot \nabla h \mathrm{d}x = \int_{\mathbb{R}^d} \nabla \cdot (h p_t v_t) \mathrm{d}x - \int_{\mathbb{R}^d} h \nabla \cdot (p_t v_t) \mathrm{d}x$$由于 $h$ 是具有紧支集的测试函数，在无穷远处为 $0$。根据散度定理，全散度项的积分为 $0$ $$\int_{\mathbb{R}^d} \nabla \cdot (h p_t v_t) \mathrm{d}x = 0$$故等式变为 $$\int_{\mathbb{R}^d} h(x) \frac{\partial p_t(x)}{\partial t} \mathrm{d}x = -\int_{\mathbb{R}^d} h(x) \nabla \cdot (p_t(x) v_t(x)) \mathrm{d}x$$

将所有项移至左侧 $$\int_{\mathbb{R}^d} h(x) \left[ \frac{\partial p_t(x)}{\partial t} + \nabla \cdot (p_t(x) v_t(x)) \right] \mathrm{d}x = 0 $$由于此等式对于任何测试函数 $h \in C_c^\infty(\mathbb{R}^d)$ 都必须成立，根据变分法基本引理，括号内的项在 $\mathbb{R}^d$ 上必须恒等于 $0$ $$\frac{\partial p_t(x)}{\partial t} + \nabla \cdot (p_t(x) v_t(x)) = 0$$

我们证毕了。我认为理解符号不容易，尤其是散度。

# 边缘矢量场的形式

我们证明，在给定边缘概率路径下，以下形式的边缘矢量场可以生成这条路径 $$u_t(x) = \int u_t(x|z) \frac{p_t(x|z)p_{data}(z)}{p_t(x)} \mathrm{d}z$$

其实这是连续性方程的一个应用。我们要寻找一个 $u_t(x)$，使得边缘连续性方程的标准形式成立 $$\frac{\partial p_t(x)}{\partial t} + \nabla \cdot (p_t(x) u_t(x)) = 0 $$我们可以看到必须满足总通量等于条件通量之和 $$p_t(x) u_t(x) = \int u_t(x|z) p_t(x|z) p_{data}(z) \mathrm{d}z$$两边同时除以 $p_t(x)$（假设 $p_t(x) > 0$），即得 $$u_t(x) = \int u_t(x|z) \frac{p_t(x|z)p_{data}(z)}{p_t(x)} \mathrm{d}z$$

更多的，根据 Bayes 定理 $$p(z | x_t=x) = \frac{p(x_t=x | z) p(z)}{p(x_t=x)}$$
有 $$u_t(x) = \int u_t(x|z) p(z | x_t=x) \mathrm{d}z = \mathbb{E}_{z \sim p(z | x_t=x)} [u_t(x | z)]$$

我们证毕了。

# L2 距离目标函数的最优解

我们证明，对于任何形如 $\min_f \mathbb{E}[\|Y - f(X)\|^2]$ 的优化问题，其理论上的最优解是 $$f^*(x) = \mathbb{E}[Y \mid X = x]$$

展开 $$J(v) = \mathbb{E} \left[ \|Y - v(X)\|^2 \right] = \mathbb{E}_X \left[ \mathbb{E}_Y \left[ \|Y - v(X)\|^2 \mid X \right] \right]$$
了最小化 $J(v)$，我们只需要对于每一个确定的 $X = x$，最小化其条件期望部分 $$\min_{v(x)} \mathbb{E} \left[ \|Y - v(x)\|^2 \mid X = x \right]$$

定义 $g(x) = \mathbb{E}[Y \mid X = x]$。我们将项 $Y - v(x)$ 进行恒等变形，插入 $g(x)$ $$\mathbb{E} \left[ \|Y - g(x) + g(x) - v(x)\|^2 \mid X = x \right]$$

根据向量模长的平方展开公式 $\|a+b\|^2 = \|a\|^2 + \|b\|^2 + 2a^\top b$，令 $a = Y - g(x)$，$b = g(x) - v(x)$ $$\mathbb{E} \left[ \|Y - g(x)\|^2 + \|g(x) - v(x)\|^2 + 2(Y - g(x))^\top (g(x) - v(x)) \mid X = x \right]$$利用期望的线性性质展开 $$\mathbb{E} \left[ \|Y - g(x)\|^2 \mid X = x \right] + \mathbb{E} \left[ \|g(x) - v(x)\|^2 \mid X = x \right] + 2\mathbb{E} \left[ (Y - g(x))^\top (g(x) - v(x)) \mid X = x \right]$$

观察最后一项。由于 $g(x)$ 和 $v(x)$ 在给定 $X = x$ 时都是常数，将它们移出期望符号 $$2(g(x) - v(x))^\top \mathbb{E} \left[ Y - g(x) \mid X = x \right]$$根据 $g(x)$ 的定义 $$\mathbb{E} [Y - g(x) \mid X = x] = \mathbb{E}[Y \mid X = x] - g(x) = g(x) - g(x) = 0$$因此，交叉项消失。

现在目标函数简化为 $$\mathbb{E} \left[ \|Y - g(x)\|^2 \mid X = x \right] + \|g(x) - v(x)\|^2$$
第一项 $\mathbb{E} [ \|Y - g(x)\|^2 \mid X = x ]$ 不依赖函数 $v(x)$ 的选取。

因此为了使表达式最小，令第二项为 $0$ $$v(x) = g(x) = \mathbb{E}[Y \mid X = x]$$
我们证毕了。
