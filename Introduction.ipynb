# Introduction

这是前言。

本文档原本是我的图像生成学习笔记，但是如今我将其写成教程形式，希望可以帮到更多人。

阅读本教程，希望你已经掌握基本的机器学习内容，如梯度下降反向传播等等，以及基本的概率论与多元微积分知识，如期望的计算。

图像生成领域经历了众多的纷杂的变迁过程。简单来说，在现代众多神经网络架构与训练技术出现之前，图像生成的方案还在灰暗的摸索之中。在 2000 年到 2014 年之间，涌现了诸如变分自编码器 VAE 与高斯混合机 GMM 之类的早期方案。这些模型性能非常低下，只能生成简单的图案。

直到 2014 年生成式对抗网络 GAN 完全打破了这一僵局。通过编码器与解码器的对抗式训练，生成式模型第一次表现出强大的性能，达到以假乱真的效果。这也离不开同期 ResNet 等等神经网络技术的发展。

后来，图像生成领域进入了蓬勃发展的时代。2017 年 Transformer 横空出世为神经网络带来新的方案。更多的，Score Matching 与成熟的 Diffusion 模型生成范式被提出。我们开始研究使用强大的神经网络来拟合真实数据分布 $p_{data}$，这种拟合方式可以是学习真实分布的梯度，或是学习纯噪声到真实图像的去噪，或是学习噪声分布到真实图像分布迁移的矢量场。

一个重大的转变是潜空间模型 Latent Diffusion Models 的提出。一个重要的流形假设是，高位的原始像素空间充斥着冗余的空间与噪声，而真实的低维语义空间仅仅占据了极小部分。我们开始重新启用 VAE，将其作为真实像素空间与潜空间之间的同构映射，将逐步生成图像的过程转移至低维潜空间内。这一范式大大减少了计算量与显存负担，是图像生成的一大里程碑。

2022 年左右，Flow Matching 理论被提出。我们发现模型可以直接学习一个噪声分布到真实图像分布的直线矢量场，将训练过程极简化，再次带来生成范式逻辑的革命。速度场预测方案成为主流。同时，更加简洁直接的生成使得少步生成甚至一步生成成为可能。

在今天，图像生成领域的主题非常简单：减少训练计算量，减少推理计算量，尽量简单，生成尽量高质量的图像。我们不断地提出新的范式，新的方法与新的工程技巧来追逐这一主题。

本教程分专题讲解。第一个专题是我们的图像生成主线，后续的专题会按照各个领域独立地展开。我尽量保持专题之间独立性。

我撰写本教程的目的，不仅仅是讲述图像生成领域主流方案的基本原理，还包含更多前沿内容。

希望你可以跟随本教程愉悦地探索这个世界 ：）
